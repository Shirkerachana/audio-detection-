Step 17: Prepare speech segments with recording IDs

```python
import os
import soundfile as sf
import numpy as np

BASE_PATH = "/content/drive/MyDrive/AI_Interview_Audio_Proctoring"

train_path = os.path.join(
    BASE_PATH,
    "train-clean-100",
    "LibriSpeech",
    "train-clean-100"
)

speech_segments = []
speech_labels = []
speech_recording_ids = []

MAX_RECORDINGS = 930
recording_count = 0

for root, dirs, files in os.walk(train_path):

    for file in files:

        if not file.endswith(".flac"):
            continue

        audio_path = os.path.join(root, file)

        waveform, sr = sf.read(audio_path)

        if sr != 16000:
            continue

        waveform = np.asarray(waveform, dtype=np.float32)

        for start in range(0, len(waveform) - 32000 + 1, 32000):

            speech_segments.append(
                waveform[start:start + 32000]
            )

            speech_labels.append(1)
            speech_recording_ids.append(recording_count)

        recording_count += 1

        if recording_count >= MAX_RECORDINGS:
            break

    if recording_count >= MAX_RECORDINGS:
        break

speech_segments = np.asarray(speech_segments, dtype=np.float32)
speech_labels = np.asarray(speech_labels, dtype=np.int64)
speech_recording_ids = np.asarray(speech_recording_ids)

print("Speech recordings:", recording_count)
print("Speech segments:", speech_segments.shape)
print("Speech labels:", speech_labels.shape)
print("Unique labels:", np.unique(speech_labels))
```

Step 18: Split recordings into train, validation and test

```python
import numpy as np
from sklearn.model_selection import GroupShuffleSplit

def split_by_recording(X, y, groups):

    splitter1 = GroupShuffleSplit(
        n_splits=1,
        test_size=0.15,
        random_state=42
    )

    train_val_idx, test_idx = next(
        splitter1.split(X, y, groups)
    )

    splitter2 = GroupShuffleSplit(
        n_splits=1,
        test_size=0.17647,
        random_state=42
    )

    train_rel, val_rel = next(
        splitter2.split(
            X[train_val_idx],
            y[train_val_idx],
            groups[train_val_idx]
        )
    )

    train_idx = train_val_idx[train_rel]
    val_idx = train_val_idx[val_rel]

    return train_idx, val_idx, test_idx


# Split speech recordings
speech_train_idx, speech_val_idx, speech_test_idx = split_by_recording(
    speech_segments,
    speech_labels,
    speech_recording_ids
)

# Split noise recordings
noise_train_idx, noise_val_idx, noise_test_idx = split_by_recording(
    noise_segments,
    noise_labels,
    noise_recording_ids
)

print("Speech segments:")
print("Train:", len(speech_train_idx))
print("Validation:", len(speech_val_idx))
print("Test:", len(speech_test_idx))

print("\nNoise segments:")
print("Train:", len(noise_train_idx))
print("Validation:", len(noise_val_idx))
print("Test:", len(noise_test_idx))
```

Step 20: Generate Log-Mel spectrograms

```python
import torch
import torchaudio
import numpy as np

SAMPLE_RATE = 16000
N_FFT = 1024
HOP_LENGTH = 256
N_MELS = 64
BATCH_SIZE = 64

mel_transform = torchaudio.transforms.MelSpectrogram(
    sample_rate=SAMPLE_RATE,
    n_fft=N_FFT,
    hop_length=HOP_LENGTH,
    n_mels=N_MELS
)

def prepare_split(speech_idx, noise_idx):

    X_batches = []
    y_batches = []

    total = len(speech_idx) + len(noise_idx)

    # Process speech and noise in small batches
    for start in range(0, total, BATCH_SIZE):

        end = min(start + BATCH_SIZE, total)

        speech_count = min(
            end, len(speech_idx)
        ) - min(start, len(speech_idx))

        noise_start = max(0, start - len(speech_idx))
        noise_end = max(0, end - len(speech_idx))

        audio_batch = []

        if speech_count > 0:
            audio_batch.append(
                speech_segments[
                    speech_idx[start:start + speech_count]
                ]
            )

        if noise_end > noise_start:
            audio_batch.append(
                noise_segments[
                    noise_idx[noise_start:noise_end]
                ]
            )

        batch = np.concatenate(audio_batch, axis=0)

        waveform = torch.tensor(batch, dtype=torch.float32)

        with torch.no_grad():
            mel = mel_transform(waveform)
            log_mel = torch.log(mel + 1e-6)

        X_batches.append(log_mel)
        y_batches.append(
            torch.tensor(
                [1] * speech_count +
                [0] * (len(batch) - speech_count),
                dtype=torch.long
            )
        )

    X = torch.cat(X_batches)
    y = torch.cat(y_batches)

    # Shuffle features and labels together
    permutation = torch.randperm(len(y))

    return X[permutation], y[permutation]


print("Generating training spectrograms...")

X_train, y_train = prepare_split(
    speech_train_idx, noise_train_idx
)

print("Generating validation spectrograms...")

X_val, y_val = prepare_split(
    speech_val_idx, noise_val_idx
)

print("Generating testing spectrograms...")

X_test, y_test = prepare_split(
    speech_test_idx, noise_test_idx
)

print("\nSpectrogram generation completed!")
print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
```

Step 23: Create CNN DataLoaders

```python
import torch
from torch.utils.data import Dataset, DataLoader

class SpeechDataset(Dataset):

    def __init__(self, X, y):
        self.X = X.float().unsqueeze(1)
        self.y = y.long()

    def __len__(self):
        return len(self.y)

    def __getitem__(self, index):
        return self.X[index], self.y[index]


BATCH_SIZE = 32

train_dataset = SpeechDataset(X_train, y_train)
val_dataset = SpeechDataset(X_val, y_val)
test_dataset = SpeechDataset(X_test, y_test)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    pin_memory=True
)

print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Testing batches:", len(test_loader))

images, labels = next(iter(train_loader))

print("Input shape:", images.shape)
print("Target shape:", labels.shape)
```

Step 25: Create  custom CNN

```python
  import torch.nn as nn

class SpeechDetectionCNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(

            nn.Conv2d(1, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.pool = nn.AdaptiveAvgPool2d((1, 1))

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.3),
            nn.Linear(64, 2)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.pool(x)
        return self.classifier(x)


model = SpeechDetectionCNN().to(device)

print(model)

print(
    "Trainable parameters:",
    sum(p.numel() for p in model.parameters() if p.requires_grad)
)
```

Step 26: Configure training

```python
import torch.optim as optim

class_counts = torch.bincount(y_train)

class_weights = (
    len(y_train) / (2.0 * class_counts.float())
).to(device)

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)

print("Class counts:", class_counts.tolist())
print("Class weights:", class_weights.tolist())
print("Optimizer: Adam")
print("Learning rate: 0.001")
```

Step 27: Train the Speech Detection CNN

```python
import os
import copy
import torch
import torch.nn as nn

from sklearn.metrics import accuracy_score, f1_score

EPOCHS = 20
PATIENCE = 5

SAVE_DIR = os.path.join(BASE_PATH, "model")
os.makedirs(SAVE_DIR, exist_ok=True)

BEST_MODEL_PATH = os.path.join(
    SAVE_DIR, "best_speech_detection_cnn.pth"
)

history = {
    "train_loss": [],
    "val_loss": [],
    "train_accuracy": [],
    "val_accuracy": []
}

best_val_loss = float("inf")
patience_counter = 0

for epoch in range(EPOCHS):

    # -------------------------
    # TRAINING
    # -------------------------
    model.train()

    running_loss = 0.0
    train_predictions = []
    train_targets = []

    for inputs, targets in train_loader:

        inputs = inputs.to(device, non_blocking=True)
        targets = targets.to(device, non_blocking=True)

        optimizer.zero_grad()

        outputs = model(inputs)
        loss = criterion(outputs, targets)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)

        predictions = torch.argmax(outputs, dim=1)

        train_predictions.extend(
            predictions.detach().cpu().numpy()
        )

        train_targets.extend(
            targets.detach().cpu().numpy()
        )

    train_loss = running_loss / len(train_dataset)

    train_accuracy = accuracy_score(
        train_targets, train_predictions
    )

    # -------------------------
    # VALIDATION
    # -------------------------
    model.eval()

    val_running_loss = 0.0
    val_predictions = []
    val_targets = []

    with torch.no_grad():

        for inputs, targets in val_loader:

            inputs = inputs.to(device, non_blocking=True)
            targets = targets.to(device, non_blocking=True)

            outputs = model(inputs)
            loss = criterion(outputs, targets)

            val_running_loss += loss.item() * inputs.size(0)

            predictions = torch.argmax(outputs, dim=1)

            val_predictions.extend(
                predictions.cpu().numpy()
            )

            val_targets.extend(
                targets.cpu().numpy()
            )

    val_loss = val_running_loss / len(val_dataset)

    val_accuracy = accuracy_score(
        val_targets, val_predictions
    )

    # Save history
    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_accuracy"].append(train_accuracy)
    history["val_accuracy"].append(val_accuracy)

    print(
        f"Epoch [{epoch + 1}/{EPOCHS}] "
        f"| Train Loss: {train_loss:.4f} "
        f"| Val Loss: {val_loss:.4f} "
        f"| Train Acc: {train_accuracy:.4f} "
        f"| Val Acc: {val_accuracy:.4f}"
    )

    # Save best model
    if val_loss < best_val_loss:

        best_val_loss = val_loss
        patience_counter = 0

        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "val_loss": val_loss,
            "history": history
        }, BEST_MODEL_PATH)

        print("Best model saved!")

    else:
        patience_counter += 1

    # Early stopping
    if patience_counter >= PATIENCE:
        print("\nEarly stopping triggered!")
        break

print("\nTraining completed!")
print("Best validation loss:", best_val_loss)
print("Best model path:", BEST_MODEL_PATH)
```

Step 30: Evaluate the best model on the test dataset

```python
import torch
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Load best checkpoint
checkpoint = torch.load(
    BEST_MODEL_PATH,
    map_location=device,
    weights_only=False
)

model.load_state_dict(checkpoint["model_state_dict"])
model.to(device)
model.eval()

print("Loaded best model from epoch:", checkpoint["epoch"])
print("Best validation loss:", checkpoint["val_loss"])

# Test evaluation
all_predictions = []
all_targets = []

with torch.no_grad():

    for inputs, targets in test_loader:

        inputs = inputs.to(device)
        outputs = model(inputs)

        predictions = torch.argmax(outputs, dim=1)

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_targets.extend(
            targets.numpy()
        )

# Calculate metrics
accuracy = accuracy_score(all_targets, all_predictions)

precision = precision_score(
    all_targets, all_predictions, zero_division=0
)

recall = recall_score(
    all_targets, all_predictions, zero_division=0
)

f1 = f1_score(
    all_targets, all_predictions, zero_division=0
)

print("\n========== TEST RESULTS ==========")
print(f"Accuracy:  {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall:    {recall * 100:.2f}%")
print(f"F1 Score:  {f1 * 100:.2f}%")

print("\nClassification Report:")
print(
    classification_report(
        all_targets,
        all_predictions,
        target_names=["Non-Speech", "Speech"],
        zero_division=0
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        all_targets,
        all_predictions
    )
)
```

```python
import numpy as np

SAMPLE_RATE = 16000
SEGMENT_LENGTH = 32000

unseen_speech_segments = []
unseen_speech_labels = []
unseen_recording_ids = []

for recording_id, sample in enumerate(unseen_speech_samples):

    waveform = np.asarray(
        sample["audio"]["array"],
        dtype=np.float32
    )

    sr = sample["audio"]["sampling_rate"]

    # Convert stereo to mono if necessary
    if waveform.ndim == 2:
        waveform = waveform.mean(axis=1)

    # Resample if required
    if sr != SAMPLE_RATE:
        import scipy.signal as signal

        waveform = signal.resample_poly(
            waveform,
            SAMPLE_RATE,
            sr
        )

    # Generate 2-second segments
    for start in range(
        0,
        len(waveform) - SEGMENT_LENGTH + 1,
        SEGMENT_LENGTH
    ):

        segment = waveform[
            start:start + SEGMENT_LENGTH
        ]

        unseen_speech_segments.append(segment)
        unseen_speech_labels.append(1)
        unseen_recording_ids.append(recording_id)


unseen_speech_segments = np.asarray(
    unseen_speech_segments,
    dtype=np.float32
)

unseen_speech_labels = np.asarray(
    unseen_speech_labels,
    dtype=np.int64
)

unseen_recording_ids = np.asarray(
    unseen_recording_ids
)

print("Unseen speech segments:", unseen_speech_segments.shape)
print("Labels:", unseen_speech_labels.shape)
print("Unique labels:", np.unique(unseen_speech_labels))
print("Unique recordings:", len(np.unique(unseen_recording_ids)))
```

Step 4 — Implement the improved training loop

```python
import os
import time
import copy
import numpy as np
import torch

# New model checkpoint path (preserve original model)
IMPROVED_MODEL_PATH = os.path.join(
    BASE_PATH,
    "model",
    "improved_speech_detection_cnn.pth"
)

history = {
    "train_loss": [],
    "val_loss": [],
    "train_accuracy": [],
    "val_accuracy": [],
    "learning_rate": [],
    "epoch_time": []
}

best_val_loss = float("inf")
epochs_without_improvement = 0
best_model_weights = copy.deepcopy(improved_model.state_dict())

print("Training loop prepared.")
print("Checkpoint:", IMPROVED_MODEL_PATH)
```

Step 5 — Start improved CNN training

```python
from tqdm.auto import tqdm

# Calculate class weights from the training labels
class_counts = torch.bincount(y_train.long(), minlength=2).float().to(device)
class_weights = (class_counts.sum() / (2 * class_counts)).to(device)

criterion = nn.CrossEntropyLoss(weight=class_weights)

print("Class weights:", class_weights.cpu().numpy())

for epoch in range(MAX_EPOCHS):
    epoch_start = time.time()

    # -------------------------
    # Training
    # -------------------------
    improved_model.train()

    train_loss_total = 0.0
    train_correct = 0
    train_total = 0

    for inputs, labels in tqdm(train_loader, desc=f"Epoch {epoch+1}/{MAX_EPOCHS}"):
        inputs = inputs.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = improved_model(inputs)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        train_loss_total += loss.item() * labels.size(0)
        predictions_batch = outputs.argmax(dim=1)

        train_correct += (predictions_batch == labels).sum().item()
        train_total += labels.size(0)

    train_loss = train_loss_total / train_total
    train_accuracy = train_correct / train_total

    # -------------------------
    # Validation
    # -------------------------
    improved_model.eval()

    val_loss_total = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = improved_model(inputs)
            loss = criterion(outputs, labels)

            val_loss_total += loss.item() * labels.size(0)
            predictions_batch = outputs.argmax(dim=1)

            val_correct += (predictions_batch == labels).sum().item()
            val_total += labels.size(0)

    val_loss = val_loss_total / val_total
    val_accuracy = val_correct / val_total

    # -------------------------
    # Learning rate scheduler
    # -------------------------
    scheduler.step(val_loss)
    current_lr = optimizer.param_groups[0]["lr"]

    # -------------------------
    # Save history
    # -------------------------
    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_accuracy"].append(train_accuracy)
    history["val_accuracy"].append(val_accuracy)
    history["learning_rate"].append(current_lr)

    epoch_time = time.time() - epoch_start
    history["epoch_time"].append(epoch_time)

    # -------------------------
    # Checkpoint & early stopping
    # -------------------------
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        epochs_without_improvement = 0

        best_model_weights = copy.deepcopy(improved_model.state_dict())

        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": best_model_weights,
            "optimizer_state_dict": optimizer.state_dict(),
            "val_loss": best_val_loss,
            "history": history
        }, IMPROVED_MODEL_PATH)

        status = "Best model saved"
    else:
        epochs_without_improvement += 1
        status = f"No improvement ({epochs_without_improvement}/{EARLY_STOPPING_PATIENCE})"

    # -------------------------
    # Epoch summary
    # -------------------------
    print(
        f"\nEpoch [{epoch+1}/{MAX_EPOCHS}] "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy*100:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy*100:.2f}% | "
        f"LR: {current_lr:.6f} | "
        f"Time: {epoch_time:.1f}s | {status}"
    )

    # Early stopping
    if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:
        print("\nEarly stopping triggered!")
        break

# Restore best model weights
improved_model.load_state_dict(best_model_weights)
improved_model.eval()

print("\nImproved training completed!")
print("Best validation loss:", best_val_loss)
print("Best model saved at:", IMPROVED_MODEL_PATH)
```
